```python
!pip install -q diffusers transformers accelerate safetensors
import torch
from diffusers import StableDiffusionPipeline
from IPython.display import display
```

```python
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
        torch_dtype=torch.float16
        )

pipe = pipe.to("cuda")
```

```python
prompt = """
A realistic portrait of a young indian woman with long dark hair,
clear natural skin, soft brown eyes, wearing a simple elegant white blouse,
subtle makeup, warm natural lighting, detailed facial features,
photorealistic photography.
"""
```

```python
image = pipe(
      prompt,
      num_images_per_prompt=1,
      num_inference_steps=30,
      guidance_scale=7.5
).images[0]
display(image)
image.save("generated_human_face.png")


```
