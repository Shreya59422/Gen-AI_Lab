```python
from transformers import pipeline

model_name = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

classifier = pipeline(
    "sentiment-analysis",
        model=model_name
        )

print("Model loaded successfully!")
```

```python
text = "I really enjoyed this product. It is excellent!"

result = classifier(text)

print("Text:", text)
print("Sentiment:", result[0]["label"])
print("Confidence:", result[0]["score"])
```

```python
texts = [
      "I really enjoyed this product.",
          "The movie was excellent and entertaining.",
              "I am very disappointed with the service.",
                  "The product quality is terrible.",
                      "The experience was amazing.",
                          "The food was okay, but nothing special."
                          ]

results = classifier(texts)

for text, result in zip(texts, results):
                              print("Text:", text)
print("Sentiment:", result["label"])
print("Confidence:", round(result["score"], 4))
print("-" * 50)
```

```python
reviews = [
      "This phone has an amazing camera.",
          "The phone battery is terrible.",
              "I loved the display of this phone.",
                  "The laptop performance is excellent.",
                      "This phone is very slow.",
                          "The headphones have excellent sound quality."
                          ]

topic = "phone"

filtered_reviews = [
                              review for review in reviews
                                  if topic.lower() in review.lower()
                                  ]

print("Topic:", topic)
print("\nFiltered Reviews:")

for review in filtered_reviews:
                      print("-", review)
```

```python
results = classifier(filtered_reviews)

print("Sentiment Analysis of Phone Reviews\n")

for review, result in zip(filtered_reviews, results):
    print("Review:", review)
    print("Sentiment:", result["label"])
    print("Confidence:", round(result["score"], 4))
    print("-" * 60)
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)

print("Tokenizer and model loaded successfully!")
```

```python
text = "This phone has an amazing camera."

inputs = tokenizer(
    text,
        return_tensors="pt"
        )

with torch.no_grad():
            outputs = model(**inputs)

print("Model Output:")
print(outputs.logits)
```

```python
predicted_class_id = outputs.logits.argmax().item()

predicted_label = model.config.id2label[predicted_class_id]

print("Text:", text)
print("Predicted Sentiment:", predicted_label)
```

```python
probabilities = torch.softmax(outputs.logits, dim=1)

predicted_class_id = probabilities.argmax().item()
predicted_label = model.config.id2label[predicted_class_id]
confidence = probabilities[0][predicted_class_id].item()

print("Text:", text)
print("Predicted Sentiment:", predicted_label)
print("Confidence:", round(confidence, 4))
```
